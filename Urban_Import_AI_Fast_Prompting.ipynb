# 1. Introducción

## 1.1 Problema a abordar

En una empresa dedicada a la importación y comercialización de
productos, los clientes pueden realizar consultas relacionadas con
productos, precios, disponibilidad y características.

Responder estas consultas de manera rápida, clara y consistente
puede representar un desafío cuando se utilizan modelos de
inteligencia artificial sin instrucciones suficientemente
estructuradas.

Un prompt demasiado general puede producir respuestas extensas,
poco precisas o incluir información que no fue proporcionada.

Por este motivo, Urban Import AI propone analizar cómo las técnicas
de Fast Prompting pueden mejorar la interacción entre un usuario y
un modelo de inteligencia artificial.

## 1.2 Relevancia

La aplicación de técnicas de prompting permite establecer con mayor
precisión el rol del modelo, el contexto de la consulta, las reglas
que debe seguir y el formato esperado de la respuesta.

Esto puede contribuir a obtener respuestas más consistentes y
reducir consultas innecesarias a la API.

## 1.3 Propuesta de solución

La propuesta consiste en desarrollar y comparar dos estrategias:

1. Un prompt básico, con instrucciones generales.
2. Un prompt optimizado utilizando técnicas de Fast Prompting.

Posteriormente se analizarán las diferencias entre ambas estrategias
y se evaluará cuál resulta más adecuada para Urban Import AI.

## 5.1 Prompt básico

El primer prompt utiliza una instrucción general para responder las

### Prompt

## 5.2 Prompt optimizado mediante Fast Prompting

El segundo prompt incorpora una estructura más detallada.

Se define el rol del asistente, el contexto, el objetivo, las
restricciones y el formato esperado de respuesta.

### Prompt optimizado

> **ROL:**  
> Sos el asistente virtual de Urban Import AI, especializado en
> atención al cliente y productos de importación.
>
> **OBJETIVO:**  
> Responder las consultas de los clientes de manera clara, breve,
> precisa y cordial.
>
> **CONTEXTO:**  
> El cliente puede consultar sobre productos, características,
> disponibilidad y precios.
>
> **REGLAS:**  
> 1. Utilizá únicamente la información proporcionada.
> 2. No inventes precios, stock ni características.
> 3. Si falta información, indicá que no está disponible.
> 4. No agregues información que no haya sido solicitada.
> 5. Utilizá un lenguaje sencillo y profesional.
>
> **FORMATO:**  
> Presentá la respuesta de forma clara y organizada.
>
> **CONSULTA DEL CLIENTE:**  
> [consulta]
> 

```python
# Prompt básico:
# contiene instrucciones generales para observar cómo responde
# el asistente cuando tiene poca información sobre el comportamiento esperado.
prompt_basico = """
Sos un asistente de Urban Import AI.
Respondé las consultas de los clientes sobre productos de importación
de forma clara y amable.
"""

# Prompt optimizado:
# incorpora técnicas de Fast Prompting como rol, objetivo,
# contexto, reglas, restricciones y formato de respuesta.
prompt_optimizado = """
ROL:
Sos el asistente virtual de Urban Import AI, especializado en
atención al cliente y productos de importación.

OBJETIVO:
Responder las consultas de los clientes de manera clara, breve,
precisa y cordial.

CONTEXTO:
El cliente puede consultar sobre productos, características,
disponibilidad y precios.

REGLAS:
1. Utilizá únicamente la información proporcionada.
2. No inventes precios, stock ni características.
3. Si falta información, indicá que no está disponible.
4. No agregues información que no haya sido solicitada.
5. Utilizá un lenguaje sencillo y profesional.

FORMATO:
Presentá la respuesta de forma clara y organizada.
"""

# Mostramos ambos prompts para poder compararlos visualmente
# antes de realizar la simulación.
print("PROMPT BÁSICO")
print("=" * 50)
print(prompt_basico)

print("\nPROMPT OPTIMIZADO")
print("=" * 50)
print(prompt_optimizado)
```

```python
# Datos de prueba del producto.
# Estos datos simulan la información que tendría disponible
# un asistente de atención al cliente.
producto = {
    "nombre": "Auriculares Bluetooth Urban X1",
    "precio": "$45.000",
    "stock": "Disponible",
    "caracteristicas": [
        "Bluetooth 5.3",
        "Batería de hasta 20 horas",
        "Micrófono integrado"
    ]
}

# Consulta que realizará el cliente al asistente.
consulta = "¿Cuánto cuestan los auriculares Urban X1 y tienen stock?"
```

## 8.1 Medición del tamaño de los prompts

Para analizar de forma más concreta el impacto de las instrucciones
utilizadas, se comparará la cantidad de palabras y caracteres presentes
en el prompt básico y en el prompt optimizado.

Esta medición permite realizar una aproximación al volumen de
información que debería procesar un modelo de inteligencia artificial.

Es importante aclarar que la cantidad de caracteres o palabras no
equivale exactamente a la cantidad de tokens. Sin embargo, permite
realizar una primera comparación cuantitativa entre ambas estrategias.

```python
# Medimos la cantidad de caracteres y palabras de cada prompt.
# Esta comparación permite observar la diferencia de tamaño
# entre el prompt básico y el prompt optimizado.

caracteres_basico = len(prompt_basico)
palabras_basico = len(prompt_basico.split())

caracteres_optimizado = len(prompt_optimizado)
palabras_optimizado = len(prompt_optimizado.split())

print("PROMPT BÁSICO")
print("Caracteres:", caracteres_basico)
print("Palabras:", palabras_basico)

print("\nPROMPT OPTIMIZADO")
print("Caracteres:", caracteres_optimizado)
print("Palabras:", palabras_optimizado)
```

## 8.2 Interpretación de la medición

La medición permite observar que el prompt optimizado contiene más
información que el prompt básico debido a que incorpora contexto,
objetivos, reglas y restricciones.

Esto demuestra que optimizar un prompt no significa necesariamente
hacerlo más corto.

El objetivo de Fast Prompting es utilizar instrucciones claras y
estructuradas para obtener mejores resultados y evitar consultas
innecesarias o respuestas que requieran nuevas interacciones.

Por este motivo, el análisis de eficiencia debe considerar tanto el

Por este motivo, en esta POC se analiza el tamaño de los prompts y se

## 9.1 Medición del tamaño de los prompts

Para analizar de forma más concreta el impacto de las instrucciones
utilizadas, se comparará la cantidad de palabras y caracteres presentes
en el prompt básico y en el prompt optimizado.

Esta medición permite realizar una aproximación al volumen de
información que debería procesar un modelo de inteligencia artificial.

Es importante aclarar que la cantidad de caracteres o palabras no
equivale exactamente a la cantidad de tokens. Por este motivo, también
se realizará una medición local de tokens.



## 9.3 Medición de tokens

Además de comparar caracteres y palabras, se realiza una estimación
del número de tokens utilizados por cada prompt.

La cantidad de tokens es una medida más relacionada con el procesamiento
de texto de los modelos de inteligencia artificial y permite realizar
un análisis más preciso del posible consumo de recursos.

Para esta prueba se utiliza la librería `tiktoken`, que permite
realizar una tokenización local sin necesidad de realizar llamadas
a una API.

```python
import tiktoken

# Utilizamos una codificación compatible con modelos de OpenAI.
encoding = tiktoken.get_encoding("cl100k_base")

# Contamos los tokens de cada prompt.
tokens_basico = len(encoding.encode(prompt_basico))
tokens_optimizado = len(encoding.encode(prompt_optimizado))

print("PROMPT BÁSICO")
print("Tokens:", tokens_basico)

print("\nPROMPT OPTIMIZADO")
print("Tokens:", tokens_optimizado)

# Calculamos la diferencia entre ambos prompts.
diferencia_tokens = tokens_optimizado - tokens_basico

print("\nDiferencia de tokens:", diferencia_tokens)
```

El prompt optimizado utiliza más tokens porque incorpora información

# 12. Conclusiones

El desarrollo de esta prueba de concepto permitió aplicar técnicas de
Fast Prompting a un caso de uso concreto: la atención de consultas de
clientes de Urban Import AI.

La comparación entre un prompt básico y un prompt optimizado permitió
observar la importancia de estructurar correctamente las instrucciones
proporcionadas al modelo.

La utilización de un rol definido, contexto, objetivos, reglas,
restricciones y formato de respuesta permite establecer un mayor
control sobre el comportamiento esperado del asistente.

Además, la medición del tamaño de los prompts permitió realizar una
comparación cuantitativa entre ambas estrategias. Esta medición también
permitió observar que optimizar un prompt no significa necesariamente
hacerlo más corto, sino utilizar instrucciones claras y estructuradas
para obtener mejores resultados y evitar interacciones innecesarias.

La incorporación de un modelo texto → imagen permitió ampliar la
propuesta original, demostrando que la inteligencia artificial puede
utilizarse tanto para generar respuestas de texto como para crear
contenido visual relacionado con los productos.

La POC desarrollada constituye una base para futuras mejoras, como la
conexión con una API, la incorporación de una base de datos de
productos, la medición de tokens y costos reales y la implementación de
una interfaz interactiva.
