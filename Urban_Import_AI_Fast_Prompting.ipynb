```python
from dotenv import load_dotenv
import os
from openai import OpenAI

# Cargar el .env
load_dotenv(override=True)

# Leer la variable
api_key = os.getenv("OPENAI_API_KEY")

# Comprobar sin mostrar la clave
print("¿Se encontró la clave?", bool(api_key))

if api_key:
    print("Prefijo:", api_key[:7])
    print("Últimos 4 caracteres:", api_key[-4:])
    print("Cantidad de caracteres:", len(api_key))

# Crear cliente
client = OpenAI(api_key=api_key)

print("Cliente creado correctamente.")
```

```python
prompt_basico = """
Responde la consulta de un cliente de Urban Import AI.

Consulta:
¿Tienen disponible el producto y cuánto cuesta?
"""

print(prompt_basico)
```

```python
prompt_optimizado = """
ROL:
Eres un asistente virtual de Urban Import AI.

OBJETIVO:
Responder consultas de clientes de manera clara, breve y profesional.

REGLAS:
- Utiliza únicamente la información proporcionada.
- No inventes precios, stock ni características de productos.
- Si falta información, indícalo claramente.
- No des información que no haya sido proporcionada.
- Mantén un tono cordial y profesional.

FORMATO DE RESPUESTA:
1. Respuesta directa.
2. Información disponible.
3. Próximo paso recomendado.

CONSULTA DEL CLIENTE:
¿Tienen disponible el producto y cuánto cuesta?
"""

print(prompt_optimizado)
```

```python
respuesta_basica = """
Sí, tenemos productos disponibles. El precio depende del producto.
"""

respuesta_optimizada = """
Para confirmar la disponibilidad y el precio necesito conocer qué
producto estás consultando.

Próximo paso:
Indicame el nombre o código del producto para poder darte una
respuesta precisa.
"""

print("=== RESPUESTA PROMPT BÁSICO ===")
print(respuesta_basica)

print("\n=== RESPUESTA PROMPT OPTIMIZADO ===")
print(respuesta_optimizada)
```

```python
criterios = [
    "Claridad",
    "Control de información",
    "Estructura",
    "Prevención de información inventada"
]

resultado_basico = [2, 1, 1, 1]
resultado_optimizado = [3, 3, 3, 3]

print("COMPARACIÓN DE PROMPTS")
print("-" * 40)

for i in range(len(criterios)):
    print(
        f"{criterios[i]}: "
        f"Básico = {resultado_basico[i]}/3 | "
        f"Optimizado = {resultado_optimizado[i]}/3"
    )
```

## Análisis de resultados

La comparación muestra que el prompt optimizado permite establecer
un comportamiento más controlado para el asistente de Urban Import AI.

La incorporación de un rol, un objetivo, reglas específicas y un
formato de respuesta permite reducir la ambigüedad y evitar que el
modelo invente información que no fue proporcionada.

Además, la estrategia busca reducir consultas innecesarias a la API,
ya que un prompt más completo puede concentrar en una única solicitud
las instrucciones necesarias para generar la respuesta.

## Análisis de eficiencia y costos

Una de las características principales de la propuesta es reducir la
cantidad de consultas realizadas al modelo.

Cada consulta a una API puede generar un costo. Por este motivo, se
busca diseñar prompts completos y eficientes que permitan obtener una
respuesta útil en una única interacción.

En esta POC se comparan dos estrategias:

- Prompt básico: requiere una interacción para obtener una respuesta,
  pero proporciona pocas instrucciones al modelo.
- Prompt optimizado: concentra el rol, objetivo, reglas y formato de
  respuesta en una única consulta.

El objetivo no es realizar más consultas, sino mejorar la calidad de
la respuesta utilizando una instrucción mejor diseñada.

```python
# Simulación de cantidad de consultas

consultas_prompt_basico = 1
consultas_prompt_optimizado = 1

print("Consultas con prompt básico:", consultas_prompt_basico)
print("Consultas con prompt optimizado:", consultas_prompt_optimizado)

if consultas_prompt_optimizado <= consultas_prompt_basico:
    print("\nEl prompt optimizado no aumenta la cantidad de consultas.")
    print("La mejora se obtiene mediante una mejor construcción del prompt.")
```

```python
# Evaluación de eficiencia

puntaje_basico = sum(resultado_basico)
puntaje_optimizado = sum(resultado_optimizado)

maximo = len(criterios) * 3

eficiencia_basico = (puntaje_basico / maximo) * 100
eficiencia_optimizado = (puntaje_optimizado / maximo) * 100

print(f"Eficiencia prompt básico: {eficiencia_basico:.1f}%")
print(f"Eficiencia prompt optimizado: {eficiencia_optimizado:.1f}%")
```

## Conclusión de la POC

La experimentación realizada permite observar que una mejor
estructuración del prompt puede mejorar la calidad y el control de las
respuestas sin aumentar la cantidad de consultas realizadas.

En el caso de Urban Import AI, el prompt optimizado incorpora un rol,
un objetivo, reglas y un formato de salida. Esto permite reducir la
ambigüedad y orientar al modelo hacia respuestas más consistentes.

La propuesta también contempla la eficiencia económica: en lugar de
realizar múltiples consultas para corregir una respuesta, se busca
concentrar las instrucciones necesarias en una única interacción.

Como siguiente etapa, esta POC podría validarse mediante llamadas
reales a una API y comparar los resultados obtenidos con los resultados
simulados en esta prueba.

# Implementación

La implementación se divide en diferentes etapas.

Primero se define una consulta representativa de un cliente de Urban
Import AI. Luego se construyen dos alternativas de prompt: una versión
básica y una versión optimizada.

La versión optimizada incorpora instrucciones específicas sobre el rol
del asistente, el objetivo de la respuesta, las restricciones y el
formato esperado.

En una implementación completa, este prompt sería enviado a un modelo
de lenguaje mediante una API. Para esta POC se utiliza una simulación
local debido a que no se realizan llamadas a una API durante las
pruebas, permitiendo analizar la metodología sin generar costos.

La estructura está preparada para incorporar posteriormente una
llamada real a la API.

```python
def construir_prompt(consulta):
    prompt = f"""
ROL:
Eres un asistente virtual de Urban Import AI.

OBJETIVO:
Responder consultas de clientes de manera clara, breve y profesional.

REGLAS:
- Utiliza únicamente información disponible.
- No inventes precios, stock ni características.
- Si falta información, indícalo claramente.
- Mantén un tono cordial y profesional.

FORMATO:
- Respuesta directa.
- Información disponible.
- Próximo paso recomendado.

CONSULTA DEL CLIENTE:
{consulta}
"""
    return prompt


consulta = "¿Tienen disponible una Toyota Hilux?"

prompt = construir_prompt(consulta)

print(prompt)
```

# Resultados

La comparación realizada muestra una mejora en la estructura del prompt
optimizado respecto del prompt básico.

El prompt optimizado permite establecer claramente el rol del asistente,
el objetivo, las reglas y el formato de respuesta.

Esto permite reducir la ambigüedad y establecer límites sobre la
información que puede utilizar el asistente.

Además, la estructura reutilizable permite procesar diferentes
consultas sin tener que construir manualmente un nuevo prompt para
cada caso.

La estrategia también favorece la eficiencia, ya que concentra las
instrucciones necesarias en una única interacción con el modelo.
